# Single Neuron

![Single Neuron](assets/neuron.png)

```python
x = [0.45, 2.35, 9.62]

w = 1.5

b = 3.0

neuron = (w * x[0]) + (w * x[1]) + (w * x[2]) + b

print(neuron)
```

# Layer of Neurons

![Layer of Neurons](assets/layer_of_neurons.png)

```python
x = [0.45, 2.35, 9.62]

w = [1.5, 2.6, 5.8]

b = [3.0, 2.0, 6.0]

layer = [w[0] * (x[0] + x[1] + x[2]) + b[0], w[1] * (x[0] + x[1] + x[2]) + b[1], w[2] * (x[0] + x[1] + x[2]) + b[2]]

print(layer)
```
